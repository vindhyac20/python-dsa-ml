```python
from collections import defaultdict
d = {"a":1, "b":2, "c":1}
groups = defaultdict(list)
for key, value in d.items():
    groups[value].append(key)
print(groups)
```

```python
s1 = "ab"
s2 = "eidbaooo"

k = len(s1)
s1_count = {}
for c in s1:
    s1_count[c] = s1_count.get(c, 0) + 1

window = {}
for c in s2[:k]:
    window[c] = window.get(c, 0) + 1

if window == s1_count:
    print(True)

for i in range(k, len(s2)):
    # add new right
    window[s2[i]] = window.get(s2[i], 0) + 1
    # drop old left
    left = s2[i - k]
    window[left] -= 1
    if window[left] == 0:
        del window[left]
    # check
    if window == s1_count:
        print(True)

print(False)
```
