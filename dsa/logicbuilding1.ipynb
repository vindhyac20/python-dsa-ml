```python
#Even or Odd – (Subtract 2 Again & Again using for loop)
n=4
for i in range(n):
    n=n-2
    if n==0:
        print("even")
        break
    if n==1:
        print("odd")
        break
```

```python
#Vowel or Consonant – Check whether a given alphabet is vowel or consonant
vowels= 'a,e,i,o,u'
alpha='u'
if alpha in vowels:
    print('vowel')
else:
    print('consonant')
```

```python
#Leap Year – Check whether a year is a leap year or not.
year=2024
if (year % 4 == 0)&(year % 100!=0):
    print("leap")
else:
    print("not leap")
```
