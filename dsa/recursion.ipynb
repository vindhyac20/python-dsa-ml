```python
def find(n):
    if n==0:
        return;
    print((n%10)+find(n//10))
```

```python
#fibonacci series
def fib(self, n: int) -> int:
        if n<=1:
            return 
        return self.fib (n-1)+self.fib(n-2)
    
```

```python
def fibonacci(n):
    if n <= 1:
        return n
    return fibonacci(n-1) + fibonacci(n-2)
fibonacci(4)
```
