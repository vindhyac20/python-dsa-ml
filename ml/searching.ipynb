```python
#searching( linear search)
arr=[1,4,7,8,9]
n=len(arr)
target=9
for i in range(0,n):
    if arr[i]==target:
        print('found at',i)
    
```

```python
#binary search
n = [1, 3, 7, 8, 12]
target = 8

low = 0
high = len(n) - 1

mid = (low + high) // 2

print(mid)

if n[mid] == target:
    print('found')
elif n[mid] < target:
    low = mid + 1
else:
    high = mid - 1
```
