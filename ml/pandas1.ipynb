```python
Get Ravi's row using .loc (by label).
Get Ravi's row using .iloc (by position) — figure out what position he's at.
Get the salary of Meera using .loc with row+column labels.
Get rows e2 through e4 using .loc (note what gets included).
Get rows at position 1 through 3 using .iloc (note what's different from #4).
Select just the name and salary columns for all rows, using .loc.
```

```python
df = pd.DataFrame({
    'name': ['Asha', 'Ravi', 'Meera', 'Kiran', 'Divya'],
    'dept': ['ML', 'SDE', 'ML', 'Data', 'SDE'],
    'salary': [85000, 72000, 91000, 68000, 75000]
}, index=['e1', 'e2', 'e3', 'e4', 'e5'])
print(df.loc['e2'])
print(df.iloc[1])
print(df.loc['e3','salary'])
print(df.loc["e2":"e4"])   #e4 also included
print(df.iloc[1:3])   # index 3 not included
print(df.loc[:,['name','salary']]) 
```
