```python
import pandas as pd
df = pd.DataFrame({
    'name': ['Asha', 'Ravi', 'Meera', 'Kiran', 'Divya', 'Suresh'],
    'dept': ['ML', 'SDE', 'ML', 'Data', 'SDE', 'ML'],
    'salary': [85000, 72000, 91000, 68000, 75000, 95000],
    'experience': [2, 4, 5, 1, 3, 6]
})
#1
df.groupby('dept')['experience'].mean()
#2
df.groupby('dept').agg(
    avg_salary=('salary','mean'),
    max_salary=('salary','max'),
    head_count=('name','count'))
#3
df.groupby('dept').size()
#4
df['dept_avg_exp']=df.groupby('dept')['experience'].transform('mean')
df['exp_vs_dept_avg']=df['experience']-df['dept_avg_exp']
print(df)
```
