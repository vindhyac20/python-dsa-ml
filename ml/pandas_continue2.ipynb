```python
#Fill missing salary using group-aware mean (per dept)
#Filter to only employees in 'ML' with experience above 2
#Print just the name and salary columns of that filtered result
```

```python
import pandas as pd
df = pd.DataFrame({
    'name': ['Zoya', 'Arjun', 'Neha', 'Farhan'],
    'dept': ['ML', 'SDE', 'ML', 'SDE'],
    'salary': [88000, None, 92000, 71000],
    'experience': [3, 5, 2, 4]
})
print(df)
```

```python
df['salary'].fillna(df.groupby('dept')['salary'].transform('mean'))
```

```python
# Just .mean() — see what shape this gives you
print(df.groupby('dept')['salary'].mean())
```

```python
# Now .transform('mean') — see how this is different
print(df.groupby('dept')['salary'].transform('mean'))
```

```python
df = pd.DataFrame({
    'name': ['Zoya', 'Arjun', 'Neha', 'Farhan'],
    'dept': ['ML', 'SDE', 'ML', 'SDE'],
    'salary': [88000, None, 92000, 71000],
    'experience': [3, 5, 2, 4]
})

df['salary'].fillna(df.groupby('dept')['salary'].transform('mean'), inplace=True)
print(df)
```

```python
#Filter to only employees in 'ML' with experience above 2
df[(df['dept']=='ML')&(df['experience']>2)]
```

```python
employees = pd.DataFrame({
    'name': ['Zoya', 'Arjun', 'Neha', 'Farhan', 'Priya'],
    'dept': ['ML', 'SDE', 'ML', 'Data', 'Design']
})

dept_info = pd.DataFrame({
    'dept': ['ML', 'SDE', 'Data'],
    'head': ['Anita', 'Suresh', 'Kavya']
})
```

```python
employees.merge(dept_info,on='dept',how='left')
```

```python
employees.merge(dept_info,on='dept',how='inner')
```

```python
employees.merge(dept_info,on='dept',how='outer')
```
