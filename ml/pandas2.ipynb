```python
Get all employees with salary above 70000.
Get all employees in 'ML' dept and salary above 80000 (use &).
Get all employees in 'ML' or 'Data' dept, using isin().
Get all employees whose name contains the letter 'i'.
Get all employees not in 'SDE' dept, using ~ and isin().
```

```python
import pandas as pd
df = pd.DataFrame({
    'name': ['Asha', 'Ravi', 'Meera', 'Kiran', 'Divya'],
    'dept': ['ML', 'SDE', 'ML', 'Data', 'SDE'],
    'salary': [85000, 72000, 91000, 68000, 75000]
}, index=['e1', 'e2', 'e3', 'e4', 'e5'])
```

```python
sal=df[df['salary']>70000]

print(sal)

and_operator=df[(df['dept']=='ML')&(df['salary']>80000)]

print(and_operator)

is_in=df[df['dept'].isin(['ML','Data'])]

print(is_in)

str_contains=df[df['name'].str.contains('i')]

print(str_contains)

isnotin=df[~df['dept'].isin(["SDE"])]
print(isnotin)
```

```python
Print how many missing values are in each column.
Print the % missing for each column.
Fill missing experience values with the median of the whole column.
Fill missing salary values using the group-aware mean pattern shown above (mean salary per dept).
After filling, confirm there are no more missing values (isna().sum() should show all zeros).
```

```python
df = pd.DataFrame({
    'name': ['Asha', 'Ravi', 'Meera', 'Kiran', 'Divya', 'Suresh'],
    'dept': ['ML', 'SDE', 'ML', 'Data', 'SDE', 'ML'],
    'salary': [85000, 72000, None, 68000, None, 95000],
    'experience': [2, None, 5, 1, 3, 6]
})
print(df.isnull().sum())
print(df.isnull().sum()/len(df))
df['experience'].fillna(df['experience'].median(),inplace=True)
df['salary'].fillna(df.groupby('dept')['salary'].transform('mean'),inplace=True)
print(df.isnull().sum())
print(df)




```
